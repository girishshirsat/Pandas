#### Group By

### groupby("column_name") - is the main function
#### Core Split-Apply-Combine Techniques

###### • agg() / aggregate() — Applies one or more summary functions to each group, either globally or mapped to specific columns using a dictionary.
###### • transform() — Applies a group-specific calculation and broadcasts the result back to the original DataFrame shape and index.
###### • filter() — Drops entire groups that do not meet a boolean condition based on group-level properties.
###### • apply() — Passes each group as a DataFrame or Series to a custom user-defined function for flexible operations.
###### • pipe() — Chains custom functions that expect a GroupBy object as input to keep multi-step pipelines readable.

#### Aggregation & Summary Statistics

###### • sum() — Computes the sum of non-null values in each group.
###### • mean() — Computes the arithmetic average of values in each group.
###### • median() — Computes the middle value (50th percentile) for each group.
###### • min() / max() — Returns the minimum or maximum value in each group.
###### • count() — Counts the number of non-null values per column in each group.
###### • size() — Returns the total number of rows in each group, including NaN values.
###### • nunique() — Counts the number of distinct values in each group.
###### • std() / var() — Computes the sample standard deviation or unbiased variance for each group.
###### • sem() — Computes the standard error of the mean for each group.
###### • skew() — Computes the unbiased skewness of values within each group.
###### • quantile() — Returns values at a specified quantile (such as 0.25 or 0.90) for each group.
###### • prod() — Computes the product of all values in each group.
###### • describe() — Generates a full table of summary statistics (count, mean, std, min, quartiles, max) for each group.
###### • value_counts() — Counts the frequency of unique row combinations within each group.
###### • corr() / cov() — Computes pairwise correlation or covariance of columns within each group.

#### Selection, Indexing & Positional Methods

###### • first() / last() — Returns the first or last non-null entry in each group.
###### • nth() — Selects the n-th row (or list of rows) from each group, including nulls by default.
###### • head() / tail() — Returns the first n or last n rows from each group.
###### • idxmax() / idxmin() — Returns the index label of the maximum or minimum value in each group.
###### • sample() — Selects a random sample of rows or a fraction of rows from each group.
###### • take() — Returns elements at specified positional indices within each group.

#### Cumulative & Window Operations

###### • cumsum() — Calculates the cumulative sum of values within each group.
###### • cumprod() — Calculates the cumulative product of values within each group.
###### • cummin() / cummax() — Tracks the running minimum or maximum value within each group.
###### • cumcount() — Numbers each item in a group sequentially starting from 0 up to group size minus one.
###### • rank() — Assigns numerical ranks to entries within each group.
###### • shift() — Shifts values forward or backward by a specified number of periods within each group.
###### • diff() — Calculates the discrete difference between consecutive elements within each group.
###### • pct_change() — Computes the percentage change between the current and prior element within each group.
###### • ffill() / bfill() — Forward-fills or backward-fills missing (NaN) values strictly within group boundaries.
###### • rolling() — Creates a rolling window object scoped independently to each group.
###### • expanding() — Creates an expanding window object scoped independently to each group.
###### • ewm() — Creates an exponentially weighted moving window object scoped to each group.

#### Group Inspection, Iteration & Utilities

###### • get_group() — Extracts a single group as a standard DataFrame or Series by its key name.
###### • groups — An attribute returning a dictionary mapping each group key to its corresponding row index labels.
###### • indices — An attribute returning a dictionary mapping each group key to integer array positions.
###### • ngroup() — Assigns a unique sequential integer ID (0 to N-1) to each group.
###### • ngroups — An attribute returning the total number of distinct groups formed.
###### • all() / any() — Checks whether all or any values in each group evaluate to True.
###### • ohlc() — Computes open, high, low, and close values for each group.
###### • resample() — Performs time-series frequency conversion and resampling independently within each group.
###### • plot() / boxplot() / hist() — Generates separate visualizations or subplots for each group.


### Types of Joins (how Parameter)

#### The how parameter specifies the type of join to perform:
##### • inner (default): Returns only rows where the join keys exist in both DataFrames.
##### • left: Returns all rows from the left DataFrame and matching rows from the right DataFrame (unmatched right values become NaN).
##### • right: Returns all rows from the right DataFrame and matching rows from the left DataFrame (unmatched left values become NaN).
##### • outer: Returns all rows from both DataFrames, filling in NaN where keys do not match.
##### • cross: Creates the Cartesian product of rows from both DataFrames (every row in left paired with every row in right).

#### Key Parameters

##### • on: The column or index name(s) to join on. Must be present in both DataFrames.
##### • left_on / right_on: Used when the join column names differ between the left and right DataFrames (e.g., left_on='emp_id', right_on='employee_id').
##### • left_index / right_index: Set to True to use the index of the left or right DataFrame as the join key instead of a column.
##### • suffixes: A tuple of string suffixes (default ('_x', '_y')) to append to overlapping column names that are not used as join keys.
##### • indicator: If True, adds a column named _merge to the output DataFrame indicating the source of each row (left_only, right_only, or both).
##### • validate: Checks merge cardinality (e.g., one_to_one, one_to_many, many_to_one, many_to_many) and raises a MergeError if violated.

#### merge() vs. concat() vs. join()

##### • pd.merge(): Combines DataFrames on shared columns or indices, similar to SQL JOIN statements.
##### • pd.concat(): Stacks DataFrames vertically (axis=0) or horizontally (axis=1) along an axis.
##### • DataFrame.join(): A shortcut for index-based merging (defaults to a left join on indices).
