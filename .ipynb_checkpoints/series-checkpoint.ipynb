```python
age=[38,63,71,42]
persons=["Tejas", "Mahesh", "Shekhar", "yash"]
#without explicit index
D=pd.Series(age)
print("witout custom index")
print(D)


print()

print("with custom index")
#with explecit index
N=pd.Series(age, index=persons)
print(N)

print()

#Assign Names
print("Assign specific name to a series")
E=pd.Series(age,index=persons,name="Age and name of persons")
print(E)
```
