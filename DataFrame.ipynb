```python
import pandas as pd
```

```python
Dct = [
    {"No": 1, "Name": "Ashish", "Age": 45},
    {"No": 2, "Name": "Nilesh", "Age": 32},
    {"No": 3, "Name": "Vyankat", "Age": 28},
    {"No": 4, "Name": "Anand", "Age": 25},
    {"No": 5, "Name": "Sumit", "Age": 58}
]
```

```python
df=pd.DataFrame(Dct,index=[row["No"] for row in Dct],columns=["Name","Age"])
#here we can also add index explicitly but here no need as  our dct is a list and contain good column
# pd.DataFrame(Dct,index=[1,2,3,4,5],columns=["No","Name","Age"])

df
```
